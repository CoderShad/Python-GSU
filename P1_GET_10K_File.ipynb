```python
# Given CIK number and year this code will get the annual report txt file 
# in the Edgar 10-K filing
#import urllib2
import time
import csv
import sys
```

```python
CIK = '0001018724'
Year = '2013'


string_match1 = 'edgar/data/1018724/0001193125-13-028520.txt'


url3 = 'https://www.sec.gov/Archives/'+string_match1 
```

```python
url3
```

```python

#words is a list

words = ['anticipate','believe', 'depend', 'fluctuate', 'indefinite', 'likelihood', 'possible', 'predict', 'risk', 'uncertain']

count={} # is a dictionary
for elem in words:
	count[elem] = 0 
```

```python
#The method split() returns a list of all the words in the string
#The split function splits a single string into a string array using the separator defined.
#If no separator is defined, whitespace is used.


## REF: https://docs.python.org/3/tutorial/stdlib.html#internet-access
    
from urllib.request import urlopen
with urlopen(url3) as response3:
    for line in response3:
        line = line.decode('utf-8')
        elements = line.split()
        for word in words:
            count[word] = count[word]+elements.count(word)

print (CIK)
print (Year)
print (url3)
print (count)


```
